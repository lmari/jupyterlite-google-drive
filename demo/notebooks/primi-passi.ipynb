# Primi passi con Python e Google Drive

Questo notebook è un esercizio guidato. Non richiede installazione di Python. Usa solo dati inventati e una copia di prova: l'integrazione con Drive è ancora sperimentale.

## 1. Esegui una cella

Seleziona la cella di codice e premi Ctrl+Invio. Il primo avvio può richiedere più tempo.

````python
import statistics as st
campione = [2, 4, 6, 8, 10]
st.mean(campione)
````

## 2. Conserva il notebook

Scegli **File → Google Drive → Connect**, autorizza l'app con il tuo account abilitato e poi scegli **Open folder**. Usa **Save notebook as new Drive copy**. Aggiorna la lista con **Refresh folder** e apri il notebook appena creato. Da questo momento Ctrl+S e l'autosalvataggio agiscono sulla copia remota. Controlla l'indicatore di salvataggio.

## 3. Leggi dati da Drive

Carica in Google Drive il file di esempio `dati.csv` con queste righe:

```csv
x,y
2,3
4,5
```

Nel notebook remoto, scegli **Mount data file…**, seleziona il CSV e conferma. Poi usa **Add selected dependency…** per registrare il riferimento nel notebook. Salva. La cella seguente funziona solo dopo il montaggio e se il filesystem del browser è disponibile.

````python
import csv
with open('/drive/data/dati.csv', newline='') as f:
    righe = list(csv.DictReader(f))
x = [float(r['x']) for r in righe]
st.mean(x)
````

Il risultato atteso è 3.0. I dati montati sono in sola lettura. Se il montaggio non funziona, consulta il docente; il comando avanzato **Import temporary data copy…** offre una copia temporanea, con un percorso diverso.

## 4. Esporta un risultato

La prossima cella crea un CSV temporaneo. Poi usa **Export output to Drive…** oppure **Download output locally…**. I risultati temporanei non vengono sincronizzati automaticamente.

````python
from pathlib import Path
Path('/drive/output/risultati.csv').write_text(f'media\n{st.mean(x)}\n', encoding='utf-8')
````
